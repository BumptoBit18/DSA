### RED-BLACK TREE

# Bài Thuyết Trình: Cây Đỏ-Đen (Red-Black Tree)

## 1. Bài toán, Đầu vào và Đầu ra

### Thuật toán giải quyết bài toán gì?
Cây Đỏ-Đen giải quyết bài toán quản lý tập dữ liệu động với các thao tác tìm kiếm, thêm mới, xóa và các truy vấn thứ tự.
Thuật toán hướng tới mục tiêu duy trì một cây tìm kiếm nhị phân tự cân bằng, khống chế chiều cao cây luôn ở mức logarit để đảm bảo độ phức tạp thời gian cho mọi thao tác trong trường hợp xấu nhất luôn là $O(\log n)$.

### Đầu vào và Đầu ra
* **Đầu vào:** Tập hợp các phần tử dữ liệu có khả năng so sánh thứ tự cùng lệnh thực thi đi kèm gồm tìm kiếm, thêm, xóa hoặc truy vấn thứ tự.
* **Đầu ra:** Kết quả tương ứng với truy vấn hoặc một cấu trúc cây Đỏ-Đen đã được tự động cân bằng lại, bảo toàn các tính chất bất biến sau mỗi thao tác.

### Bảng tổng hợp các thao tác cơ bản

| Thao tác | Đầu vào | Đầu ra | Độ phức tạp |
| :--- | :--- | :--- | :--- |
| **Tìm kiếm** | Cây Đỏ-Đen $T$ và khóa $k$ | Nút chứa $k$ hoặc `NIL` | $O(\log n)$ |
| **Thêm mới** | Cây Đỏ-Đen $T$ và khóa $k$ | Cây Đỏ-Đen $T$ đã thêm $k$ và cân bằng | $O(\log n)$ |
| **Xóa** | Cây Đỏ-Đen $T$ và khóa $k$ | Cây Đỏ-Đen $T'$ đã loại bỏ $k$ và cân bằng | $O(\log n)$ |
| **Min, Max, Successor, Predecessor** | Cây Đỏ-Đen $T$ và một nút $x$ | Nút tương ứng | $O(\log n)$ |

## 2. Nguyên lý và Chiến lược của Thuật toán

### Nguyên lý của thuật toán
* **Mô phỏng Cây 2-3 trên cấu trúc nhị phân:** Thuật toán sử dụng nền tảng của cây tìm kiếm nhị phân cố định, bổ sung một bit lưu thông tin màu đỏ hoặc đen cho mỗi nút. Khối nút gồm nút Đen nối với nút Đỏ đóng vai trò đại diện cho nút chứa hai giá trị của cây 2-3.
* **Khống chế chiều cao bằng 5 quy tắc bất biến:** Tính cân bằng xấp xỉ của cây được duy trì nghiêm ngặt thông qua 5 quy tắc:
  1. Mỗi nút chỉ mang màu đỏ hoặc đen.
  2. Nút gốc bắt buộc màu đen.
  3. Mọi nút lá rỗng `NIL` đều màu đen.
  4. Nút màu đỏ bắt buộc phải có hai nút con màu đen, nghĩa là không tồn tại hai nút đỏ nối tiếp nhau.
  5. Mọi đường đi từ một nút xuống các lá `NIL` hậu duệ đều chứa cùng một số lượng nút đen.

Đường đi dài nhất từ gốc đến lá không vượt quá hai lần đường đi ngắn nhất. Kết hợp với việc cây có chiều cao đen $bh$ chứa tối thiểu $2^{bh} - 1$ nút nội, chiều cao $h$ của cây chứa $n$ phần tử luôn thỏa mãn bất đẳng thức:
$$h \le 2 \log_2(n+1) = O(\log n)$$

### Chiến lược xử lý
Khi thực hiện thao tác thêm hoặc xóa, thuật toán trước hết thực hiện như cây tìm kiếm nhị phân thông thường, sau đó khôi phục tính chất bất biến bằng cách lan truyền điều chỉnh cục bộ từ vị trí biến đổi ngược lên gốc:
* **Chèn:** Mặc định tô màu đỏ cho nút mới để không làm thay đổi số lượng nút đen trên các đường đi. Nếu phát sinh lỗi hai nút đỏ liên tiếp, cây sẽ khôi phục trạng thái bằng đổi màu hoặc xoay với số phép xoay tối đa là 2.
* **Xóa:** Việc xóa một nút đen làm mất cân bằng số lượng nút đen trên đường đi sẽ được khắc phục thông qua mối quan hệ màu sắc với nút anh em, với số phép xoay tối đa là 3.
* **Đổi màu:** Áp dụng khi nút cha và nút chú cùng màu đỏ, giúp xử lý điểm quá tải bằng cách đẩy giá trị lên tầng trên.
* **Xoay cây:** Gồm xoay trái và xoay phải, áp dụng khi nút cha màu đỏ nhưng nút chú màu đen, giúp tái cấu trúc lại hình dạng cây mà vẫn giữ nguyên thứ tự tìm kiếm nhị phân.

Mỗi bước đổi màu hoặc xoay cây chỉ tiêu tốn thời gian $O(1)$. Do quá trình điều chỉnh chỉ di chuyển tối đa dọc theo chiều cao của cây, tổng chi phí cho mỗi thao tác chèn hoặc xóa luôn đảm bảo tuyệt đối ở mức $O(\log n)$ trong mọi trường hợp.

## 3. Minh họa bằng tay

Ta dùng ký hiệu số `(R)` = nút đỏ, số `(B)` = nút đen. Mọi cây dưới đây khớp với kết quả chương trình in ra ở mục 4.

### 3.1. Ví dụ THÊM: lần lượt 10, 20, 30, 15, 12

* **Bước 1 – thêm 10:** Cây rỗng nên 10 làm gốc. Nút mới màu đỏ nhưng luật 2 bắt gốc phải đen $\rightarrow$ tô đen.
  ```text
  10(B)
  ```
* **Bước 2 – thêm 20:** $20 > 10$ nên sang phải, tô đỏ. Cha là 10 (đen) $\rightarrow$ không vi phạm gì.
  ```text
  10(B)
  └── 20(R)
  ```
* **Bước 3 – thêm 30:** Đi sang phải của 20, tô đỏ. 
  Giờ 20(R) – 30(R) hai đỏ liền nhau! Chú (con trái của 10) là NIL = đen, và $10 \rightarrow 20 \rightarrow 30$ thẳng hàng sang phải $\Rightarrow$ TH3: cha 20 $\rightarrow$ đen, ông 10 $\rightarrow$ đỏ, xoay trái tại 10
  ```text
  Trước khi sửa                                 |   Sau khi sửa
  10(B)                                         |      20(B)
    \                                           |     /    \
    20(R)                                       |   10(R)   30(R)
      \                                         |        
      30(R)                                     | 
  ```
* **Bước 4 – thêm 15:** $15 < 20 \rightarrow$ trái; $15 > 10 \rightarrow$ phải của 10. Cha 10 đỏ, chú 30 cũng đỏ $\Rightarrow$ TH1: 10 và 30 $\rightarrow$ đen, ông 20 $\rightarrow$ đỏ. Ông 20 là gốc nên bị ép lại thành đen.  
  ```text
      20(B)
    /       \
  10(B)     30(B)
      \
      15(R)
  ```
* **Bước 5 – thêm 12:** Đi: $20 \rightarrow$ trái $10 \rightarrow$ phải $15 \rightarrow$ trái, đặt 12 (đỏ). Cha 15 đỏ, chú (con trái của 10) là NIL đen. Đường $10 \rightarrow 15 \rightarrow 12$ gấp khúc (phải rồi trái) $\Rightarrow$ TH2 rồi TH3:
  1. TH2: xoay phải tại 15 để thành đường thẳng $10 \rightarrow 12 \rightarrow 15$.
  2. TH3: 12 (cha) $\rightarrow$ đen, 10 (ông) $\rightarrow$ đỏ, xoay trái tại 10.
  ```text
  Sau TH2                                        |      Sau TH3
  10(B)                                          |       20(B)
    \                                            |      /    \
    12(R)                                        |    12(B) 30(B)
      \                                          |     / \
      15(R)                                      |  10(R) 15(R) 
  ```
Kiểm tra 5 luật: gốc 20 đen ✔; không có đỏ nào có con đỏ ✔; mọi đường từ gốc xuống NIL đều qua đúng 2 nút đen (20 và 12, hoặc 20 và 30) ✔

### 3.2: Ví dụ XÓA: Ta dùng cây tạo bởi dãy 10, 20, 30, 15, 12, 25, 5 (thêm 25 và 5 theo cách tương tự; thêm 5 sẽ kích hoạt TH1)

  ```text
    20(B)
      ├── L: 12(R)
      │   ├── L: 10(B)
      │   │   └── L: 5(R)
      │   └── R: 15(B)
      └── R: 30(B)
          └── L: 25(R) 
  ```
* Xóa 5 (lá đỏ): Nút đỏ bị cắt đi không làm mất trạm đen nào $\rightarrow$ không cần sửa gì.
* Xóa 20 (gốc, có hai con): Lấy nút kế nhiệm = nút nhỏ nhất bên phải = 25 (đỏ) lên thay chỗ 20 và thừa hưởng màu đen của 20[cite: 2]. Nút bị “rút” khỏi chỗ cũ là 25 (màu đỏ) $\Rightarrow$ không mất trạm đen $\Rightarrow$ không cần sửa.
  ```text
    25(B)
    ├── L: 12(R)
    │   ├── L: 10(B)
    │   └── R: 15(B)
    └── R: 30(B)
  ```
* Xóa 10 (lá đen): Mất một trạm đen $\rightarrow$ vị trí trống mang “đen thừa”[cite: 2]. Anh/em của nó là 15 (đen), và hai con của 15 đều NIL (đen) $\Rightarrow$ TH2: tô 15 $\rightarrow$ đỏ, đẩy “đen thừa” lên cha là 12[cite: 2]. Nhưng 12 đang đỏ, nên chỉ cần tô 12 às đen là xong (đen thừa được hấp thụ).
  ```text
  25(B)
  ├── L: 12(B)
  │   └── R: 15(R)
  └── R: 30(B)
  ```

* Kiểm tra:
    * Đường $25 \rightarrow 12 \rightarrow NIL$: 2 nút đen.
    * Đường $25 \rightarrow 12 \rightarrow 15(R) \rightarrow NIL$: 2 nút đen (15 đỏ không tính).
    * Đường $25 \rightarrow 30 \rightarrow NIL$: 2 nút đen ✔.

## 4.1. Bản đồ của chương trình

Chương trình được xây dựng từ hai lớp chính: `Node` đại diện cho một nút trong cây và `RedBlackTree` quản lý toàn bộ cây.

Các thành phần được chia thành những nhóm chức năng sau:

| Nhóm | Thành phần | Nhiệm vụ |

| Cấu trúc | `Node`, `__init__` | Tạo nút và khởi tạo cây rỗng |
| Công cụ sửa cây | `left_rotate`, `right_rotate` | Thực hiện phép xoay trái và xoay phải |
| Thêm | `insert`, `insert_fixup` | Thêm nút mới và khôi phục các tính chất đỏ–đen |
| Tìm kiếm | `search`, `ceiling`, `floor` | Tìm kiếm và truy vấn theo thứ tự |
| Xóa | `delete`, `transplant`, `minimum`, `delete_fixup` | Xóa nút và xử lý trường hợp mất cân bằng màu |
| Tiện ích | `height`, `is_valid`, `pretty` | Duyệt cây, tính chiều cao, kiểm tra tính hợp lệ và in cấu trúc cây |

## 4.2. Lớp Node và việc khởi tạo cây

Mỗi nút lưu trữ khóa (`key`), giá trị đi kèm (`val`), màu (`color`), hai con trỏ đến con trái và con phải (`left`, `right`), cùng con trỏ đến nút cha (`parent`). Con trỏ cha giúp thuật toán dễ dàng di chuyển ngược lên các tầng phía trên khi cần sửa cây.

Thuộc tính `__slots__` giới hạn các thuộc tính được phép có trong mỗi đối tượng `Node`, góp phần tiết kiệm bộ nhớ khi cây chứa số lượng lớn nút.

Chương trình sử dụng hai hằng số `RED` và `BLACK` để biểu diễn màu sắc. Đặc biệt, thay vì dùng `None` cho tất cả các vị trí con không tồn tại, cây sử dụng một nút lính canh `NIL` duy nhất có màu đen. Các con trỏ chưa có nút thực đều trỏ đến `NIL`, giúp việc kiểm tra màu sắc và duy trì tính chất của cây trở nên nhất quán hơn.

Khi khởi tạo, `root` trỏ đến `NIL` để biểu diễn cây rỗng, còn `size` lưu số lượng nút hiện có. Tùy chọn `verbose=True` cho phép chương trình in ra thông tin về các bước sửa cây, thuận tiện cho việc theo dõi thuật toán khi thuyết trình.

**Mã nguồn:** [Xem phần khởi tạo Node](node.py)

## 4.3. Hai phép xoay

Phép xoay là công cụ quan trọng giúp cây đỏ–đen khôi phục cấu trúc cân bằng mà vẫn giữ nguyên thứ tự của cây tìm kiếm nhị phân.

- **Xoay trái:** đưa con phải của một nút lên vị trí cao hơn, đồng thời chuyển nhánh con trái của nút đó sang làm con phải của nút ban đầu.
- **Xoay phải:** thực hiện thao tác đối xứng với xoay trái.

Trong quá trình xoay, chương trình phải cập nhật chính xác các con trỏ cha, con trái, con phải và nút gốc nếu cần. Mỗi phép xoay chỉ thay đổi một số lượng hằng số các con trỏ nên có độ phức tạp thời gian \(O(1)\).

Điểm cần nhớ là phép xoay thay đổi cấu trúc cây nhưng không làm thay đổi thứ tự khóa: các khóa vẫn thỏa mãn quan hệ của cây tìm kiếm nhị phân.

**Mã nguồn:** [Xem cài đặt phép xoay trái và xoay phải](rotater.py#1)

## 4.4. Thêm một nút

Thao tác `insert(key, val)` trước tiên tìm vị trí thích hợp để chèn khóa mới theo quy tắc của cây tìm kiếm nhị phân: khóa nhỏ hơn đi sang trái, khóa lớn hơn đi sang phải.

Nếu khóa đã tồn tại, chương trình cập nhật giá trị tương ứng và trả về `False`, không tạo thêm nút mới. Nếu khóa chưa tồn tại, chương trình tạo một nút mới màu đỏ, gắn nút vào đúng vị trí và tăng số lượng phần tử của cây.

Sau khi chèn, thuật toán gọi `insert_fixup` để xử lý những vi phạm có thể phát sinh, đặc biệt là trường hợp nút mới và nút cha đều mang màu đỏ. Nếu chèn nút mới vào cây rỗng, nút đó trở thành gốc và cuối cùng được tô đen.

## 4.5. Sửa cây sau khi thêm

Đây là phần cốt lõi của thuật toán chèn. Việc sửa cây chỉ cần tiếp tục khi nút đang xét có cha màu đỏ, bởi hai nút đỏ liên tiếp vi phạm tính chất thứ tư của cây đỏ–đen.

Gọi `z` là nút đang được xử lý, `z.parent` là cha, `gp` là ông và `u` là chú của nút. Thuật toán xét màu của chú và vị trí tương đối giữa nút, cha và ông để lựa chọn trường hợp xử lý.

| Trường hợp | Điều kiện | Cách xử lý |
|---|---|---|
| TH1 – Chú đỏ | Cha và chú đều đỏ | Tô cha và chú thành đen, tô ông thành đỏ rồi tiếp tục kiểm tra từ ông |
| TH2 – Dạng tam giác | Chú đen, nút và cha tạo thành đường gấp khúc | Xoay tại cha để chuyển thành dạng đường thẳng, sau đó xử lý theo TH3 |
| TH3 – Dạng đường thẳng | Chú đen, nút, cha và ông tạo thành đường thẳng | Đổi màu cha và ông, sau đó xoay tại ông |

Các trường hợp được xử lý đối xứng khi cha nằm ở nhánh trái hoặc nhánh phải của ông. Sau khi kết thúc, thuật toán luôn ép nút gốc về màu đen để bảo đảm tính chất thứ hai.

Trong quá trình cài đặt, các thông báo được in ra khi bật `verbose=True` giúp quan sát từng lần đổi màu và xoay cây. Các thông báo này chỉ phục vụ việc theo dõi, không phải điều kiện bắt buộc để thuật toán hoạt động.

## 4.6. Tìm kiếm, ceiling và floor

Chương trình cung cấp các thao tác tìm kiếm và truy vấn theo thứ tự:

- `search(key)`: trả về nút chứa khóa cần tìm hoặc `None` nếu không tồn tại.
- `ceiling(key)`: tìm khóa nhỏ nhất thỏa mãn điều kiện lớn hơn hoặc bằng `key`.
- `floor(key)`: tìm khóa lớn nhất thỏa mãn điều kiện nhỏ hơn hoặc bằng `key`.

Hàm tìm kiếm di chuyển từ gốc xuống cây, mỗi bước lựa chọn nhánh trái hoặc phải dựa trên phép so sánh khóa. Do cây đỏ–đen được duy trì chiều cao logarithmic, thao tác tìm kiếm có độ phức tạp \(O(\log n)\).

Đối với `ceiling`, mỗi khi gặp một khóa lớn hơn hoặc bằng khóa cần tìm, thuật toán ghi nhớ nút đó làm ứng viên rồi tiếp tục sang trái để tìm một khóa nhỏ hơn nhưng vẫn thỏa mãn điều kiện. Với `floor`, cách thực hiện đối xứng: khi gặp khóa nhỏ hơn hoặc bằng khóa cần tìm, thuật toán ghi nhớ ứng viên rồi tiếp tục sang phải.

Hai thao tác `ceiling` và `floor` đặc biệt hữu ích trong các bài toán cần tìm phần tử gần nhất theo thứ tự, chẳng hạn kiểm tra khoảng thời gian có bị trùng lặp hay không.

**Mã nguồn:** [Xem cài đặt tìm kiếm, ceiling và floor](searcher.py)

## 4.7. Xóa một nút

Thao tác `delete(key)` trước hết tìm nút cần xóa. Nếu khóa không tồn tại, hàm trả về `False`. Nếu tìm thấy, thuật toán xử lý theo số lượng con của nút:

1. **Không có con trái:** lấy con phải thay thế nút cần xóa.
2. **Không có con phải:** lấy con trái thay thế nút cần xóa.
3. **Có đủ hai con:** tìm nút kế nhiệm, tức nút có khóa nhỏ nhất trong cây con phải, rồi đưa nút đó lên thay thế.

Hàm `transplant(u, v)` hỗ trợ thay thế cây con bằng cách nối `v` vào đúng vị trí trước đó của `u`. Hàm `minimum(x)` tìm nút nhỏ nhất trong cây con bắt đầu từ `x`.

Hai biến quan trọng trong quá trình xóa là `y_color`, màu của nút thực sự bị loại khỏi vị trí ban đầu, và `x`, nút thay thế vị trí đó. Nếu `y_color` là đỏ, số nút đen trên các đường đi không bị giảm nên không cần sửa thêm. Ngược lại, nếu nút bị loại là đen, thuật toán phải gọi `_delete_fixup(x)` để xử lý tình trạng mất một nút đen, thường được mô tả bằng khái niệm “đen thừa” (*double black*).

**Mã nguồn:** [Xem cài đặt thao tác xóa](deleter.py)

## 4.8. Sửa cây sau khi xóa

Khi một nút đen bị loại khỏi cây, một đường đi có thể mất đi một nút đen so với các đường đi khác. Thuật toán sửa cây sử dụng `x` để theo dõi vị trí đang mang trạng thái “đen thừa” và `w` để biểu diễn nút anh/em của `x`.

Có bốn trường hợp chính:

| Trường hợp | Điều kiện | Cách xử lý |
|---|---|---|
| TH1 | Anh/em `w` màu đỏ | Đổi màu và xoay để chuyển sang một trường hợp có anh/em màu đen |
| TH2 | `w` màu đen, cả hai con của `w` đều đen | Tô `w` đỏ và đẩy trạng thái “đen thừa” lên cha |
| TH3 | `w` màu đen, con xa màu đen, con gần màu đỏ | Đổi màu và xoay tại `w` để chuyển thành TH4 |
| TH4 | `w` màu đen, con xa màu đỏ | Đổi màu, tô đen cha và con xa, xoay tại cha rồi kết thúc |

Các trường hợp được xử lý đối xứng tùy theo `x` là con trái hay con phải. TH2 là trường hợp có thể khiến thuật toán tiếp tục đi lên nhiều tầng; những trường hợp còn lại chuyển trạng thái hoặc kết thúc việc sửa cây.

Cuối cùng, nút `x` được tô đen để hấp thụ trạng thái “đen thừa”, khôi phục các tính chất cần thiết của cây đỏ–đen.

**Mã nguồn:** [Xem cài đặt thuật toán sửa cây sau khi xóa](deleter.py#52)

## 4.9. Các hàm tiện ích

Ngoài các thao tác chính, chương trình còn cung cấp các hàm hỗ trợ kiểm tra và quan sát cấu trúc cây.

- `height()`: tính chiều cao cây. Cây rỗng có chiều cao bằng 0; chiều cao của một nút bằng 1 cộng với chiều cao lớn hơn giữa hai cây con.
- `is_valid()`: kiểm tra tính hợp lệ của cây, bao gồm gốc màu đen, thứ tự khóa của BST, không có hai nút đỏ liên tiếp, các con trỏ cha chính xác và số nút đen trên các đường đi nhất quán.
- `pretty()`: biểu diễn cấu trúc cây dưới dạng văn bản, sử dụng `L` và `R` để chỉ con trái và con phải, đồng thời hiển thị màu của từng nút.

Trong đó, `is_valid()` đóng vai trò như một bộ kiểm tra tự động sau mỗi thao tác thêm hoặc xóa. Hàm giúp phát hiện sớm các lỗi trong quá trình cài đặt và là thành phần quan trọng trong kiểm thử thuật toán.

## 6. Độ phức tạp thời gian

| Thao tác | Tốt nhất | Trung Bình | Xấu nhất |
| :--- | :--- | :--- | :--- |
| **Search** | $O(\log n)$ | $O(\log n)$ | $O(\log n)$ |
| **Insert** | $O(\log n)$ | $O(\log n)$ | $O(\log n)$ |
| **Delete** | $O(\log n)$ | $O(\log n)$ | $O(\log n)$ |
| **Rotation** | $O(1)$ | $O(1)$ | $O(1)$ |
| **Duyệt toàn bộ cây** | $O(n)$ | $O(n)$ | $O(n)$ |

### Giải thích chi tiết:
* **Tìm kiếm là $O(\log n)$ vì:**
  * Red-Black Tree tự cân bằng bằng cách tô màu các nút đỏ/đen và thực hiện phép xoay khi cần thiết.
  * Mỗi lần tìm kiếm, ta đi sang trái hoặc phải, tương tự Binary Search Tree.
  * Chiều cao cây luôn bị giới hạn bởi $h \le 2 \log_2(n+1)$.
  * Vì vậy, số bước tìm kiếm tỷ lệ với chiều cao cây, tức là $O(\log n)$.
  * *Ví dụ:* Với $n = 1.000$ nút, chiều cao cây bị giới hạn khoảng 20 tầng, nên tìm kiếm chỉ cần đi qua tối đa khoảng 20 nút.
* **Tóm tắt:**
  * `Search`: $O(\log n)$ tìm kiếm theo chiều cao cây.
  * `Insert/Delete`: $O(\log n)$ tìm vị trí và cân bằng lại cây.
  * `Rotation`: $O(1)$ chỉ thay đổi liên kết giữa một số nút.
  * $\rightarrow$ Red-Black Tree bảo đảm tìm kiếm, thêm và xóa đều có độ phức tạp $O(\log n)$ trong trường hợp xấu nhất, tốt hơn so với BST thông thường có thể suy biến thành $O(n)$.

## 7. Độ phức tạp không gian

| Thao tác | Độ phức tạp không gian |
| :--- | :--- |
| **Lưu trữ $n$ nút trong cây** | $O(n)$ |
| **Search** | $O(1)$ nếu lặp, $O(\log n)$ nếu đệ quy |
| **Insert** | $O(1)$ bộ nhớ phụ nếu cài đặt lặp |
| **Delete** | $O(1)$ bộ nhớ phụ nếu cài đặt lặp |
| **Duyệt toàn bộ cây** | $O(n)$ nếu lưu kết quả, hoặc $O(\log n)$ bộ nhớ phụ với duyệt đệ quy |

### Giải thích:
* **$O(n)$:** Cần lưu $n$ nút; mỗi nút chứa dữ liệu, con trỏ trái/phải, con trỏ cha và màu đỏ/đen[cite: 4].
* **$O(1)$:** Chỉ cần một số biến phụ cố định[cite: 4].
* **$O(\log n)$:** Ngăn xếp đệ quy tăng theo chiều cao cây; Red-Black Tree có chiều cao $O(\log n)$[cite: 4].

## 8. Bài toán ứng dụng: My Calendar I (Leetcode)

* **Link bài toán:** [Leetcode - My Calendar I](https://leetcode.com/problems/my-calendar-i/description/)

### Ý tưởng thuật toán:
* Mỗi sự kiện được biểu diễn bằng khoảng thời gian `[start, end)`.
  * *Ví dụ:*
    * `[10, 20):` đặt thành công.
    * `[15, 25):` thất bại vì trùng với `[10, 20)`.
    * `[20, 30):` đặt thành công vì sự kiện trước kết thúc đúng lúc sự kiện mới bắt đầu.
* Ta xây dựng Red-Black Tree theo quy tắc:
  * Khóa của mỗi nút: thời gian bắt đầu `start`.
  * Giá trị lưu: thời gian kết thúc `end`.
  * Các nút bên trái có `start` nhỏ hơn; các nút bên phải có `start` lớn hơn.
* Khi đặt lịch mới, tìm vị trí thích hợp trong cây và kiểm tra xem khoảng thời gian có bị trùng không. Nếu không trùng, chèn nút và cân bằng cây bằng các phép xoay, đổi màu.

### Các thành phần chính trong cài đặt:
* `book():` kiểm tra trùng lịch, chèn sự kiện mới.
* `fix():` cân bằng cây bằng đổi màu và xoay cây.
* `rotate_left()` và `rotate_right():` thay đổi cấu trúc cây để duy trì cân bằng.
* `Node:` lưu thời gian bắt đầu, kết thúc, màu sắc và các liên kết.
