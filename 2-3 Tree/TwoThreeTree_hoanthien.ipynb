## 1. Bài toán 

### 1.1. Thuật toán giải quyết bài toán gì?
* **Bài toán Bảng ký hiệu có thứ tự:** Quản lý một tập hợp dữ liệu động hỗ trợ hiệu quả các thao tác: **Tìm kiếm (`search`)**, **Chèn (`insert`)**, **Xóa (`delete`)** và truy vấn theo thứ tự.
* **Hạn chế của Cây nhị phân tìm kiếm (BST) truyền thống:** 
  * BST thông thường có độ phức tạp trung bình là $\mathcal{O}(\log N)$, nhưng phụ thuộc hoàn toàn vào thứ tự dữ liệu đầu vào.
  * Trong **trường hợp xấu nhất** (dữ liệu đầu vào đã sắp xếp tăng/giảm dần), BST bị suy biến thành danh sách liên kết đơn (chiều cao $h = N$), khiến mọi thao tác chậm xuống mức $\mathcal{O}(N)$.
* **Mục tiêu của Cây 2-3:** Đảm bảo cây **cân bằng tuyệt đối (perfectly balanced)** mọi lúc, giữ cho chiều cao cây luôn đạt $\mathcal{O}(\log N)$ trong cả trường hợp xấu nhất.

### 1.2. Đầu vào và Đầu ra
* **Đầu vào (Input):** 
  * Một dãy các khóa $K = \{k_1, k_2, \dots, k_n\}$ có thể so sánh được (Comparable keys) cùng các yêu cầu thao tác (`insert(k)`, `search(k)`).
* **Đầu ra (Output):**
  * Cấu trúc **Cây 2-3 hợp lệ** (mọi nút lá nằm trên cùng một mức, các khóa phân bố đúng thứ tự từ điển).
  * Kết quả truy vấn: trả về `True` (hoặc giá trị gắn với khóa) nếu tìm thấy, `False` (hoặc `None`) nếu không tồn tại.

## 2. Ý tưởng cốt lõi 

### 2.1. Cấu trúc nút linh hoạt (Cho phép nút "béo" lên)
Thay vì ép mỗi nút chỉ được chứa đúng 1 khóa như BST, Cây 2-3 cho phép tồn tại **2 loại nút**:

| Loại nút | Số khóa (Keys) | Số nhánh con (Links) | Quy tắc phân hoạch khoảng giá trị |
| :--- | :---: | :---: | :--- |
| **Nút 2 (2-node)** | $1$ khóa ($K_1$) | $2$ con (`left`, `right`) | $\text{left} < K_1 < \text{right}$ |
| **Nút 3 (3-node)** | $2$ khóa ($K_1 < K_2$) | $3$ con (`left`, `mid`, `right`) | $\text{left} < K_1 < \text{mid} < K_2 < \text{right}$ |

* **Bất biến cân bằng hoàn hảo (Perfect Balance Invariant):** Mọi đường đi từ nút gốc (root) đến tất cả các nút lá (`null` links) đều có **cùng độ dài**.
* Chiều cao $H$ của cây 2-3 chứa $N$ khóa luôn thỏa mãn:
  $$\log_3(N + 1) - 1 \le H \le \log_2(N + 1) - 1$$

### 2.2. Chiến lược chính: Phát triển từ dưới lên (Bottom-up Growth & Local Split)
* **Tại sao BST bị lệch?** Vì BST phát triển **từ trên xuống (Top-down)**: khóa mới luôn bị gắn thêm làm lá mới ở đáy cây, làm tăng chiều cao cục bộ.
* **Chiến lược của Cây 2-3:** Không bao giờ tạo nút lá mới ở dưới đáy!
  1. Tìm xuống nút lá phù hợp và **nhét thẳng khóa mới vào nút lá đó**.
  2. Nếu nút lá đang là **2-node** (1 khóa) $\rightarrow$ trở thành **3-node** (2 khóa). Xong!
  3. Nếu nút lá đang là **3-node** (2 khóa) $\rightarrow$ tạm thời tràn thành **4-node** (chứa 3 khóa: $S < M < L$).
  4. **Phép tách cục bộ (Local Split):** Tách 4-node tạm thời thành hai 2-node chứa $S$ và $L$, đồng thời **đẩy khóa ở giữa ($M$) lên nút cha**.
  5. Quá trình đẩy khóa giữa có thể lan truyền ngược lên tới gốc. Nếu **nút gốc** bị tràn thành 4-node và bị tách, chiều cao của toàn bộ cây mới tăng thêm $1$ bậc cùng lúc!

## 1. Insert

Ý tưởng: luôn chèn vào lá, nếu lá bị đầy thì tách và đẩy khóa giữa lên.

Đi xuống như search để tìm lá phù hợp.
Chèn khóa vào lá (giữ thứ tự).
Lá là nút 2 thì thành nút 3, xong.
Lá là nút 3 thì tạm thành nút 4 (3 khóa a < b < c), cần tách.
Tách nút 4: khóa giữa b đẩy lên cha, a và c thành hai nút 2 riêng, gắn làm hai con của cha.
Cha nhận thêm b. Nếu cha cũng thành nút 4 thì lặp lại việc tách, cứ thế lên trên.
Nếu gốc bị tách thì tạo gốc mới chứa b. Đây là cách duy nhất cây cao thêm.

Vì cây chỉ cao thêm ở gốc nên mọi lá vẫn cùng độ sâu. Độ phức tạp: O(log N).

## 2. Search

Ý tưởng: đi từ gốc xuống, mỗi nút chỉ cần so sánh khóa cần tìm với 1 hoặc 2 khóa của nút để chọn đúng một nhánh con.

* Bắt đầu ở gốc.
* Nếu khóa trùng với một khóa trong nút thì tìm thấy, dừng.
* Nếu đã tới lá mà không trùng thì khóa không tồn tại.
* Nút 2 (khóa k): nhỏ hơn k đi con trái, lớn hơn k đi con phải.
* Nút 3 (khóa a < b): nhỏ hơn a đi con trái, nằm giữa a và b đi con giữa, lớn hơn b đi con phải.

Mỗi bước đi xuống đúng một tầng và chiều cao cây là O(log N), nên độ phức tạp: O(log N).

# Tìm kiếm

**`search(35)`**
1. Gốc `[40]`: 35 < 40, đi sang con trái.
2. Nút `[20, 30]`: 35 > 30, đi sang con thứ 3.
3. Lá `[35]`: tìm thấy.

→ `True`

**`search(99)`**
1. Gốc `[40]`: 99 > 40, sang con phải.
2. Nút `[60]`: 99 > 60, sang con phải.
3. Lá `[70]`: 99 ≠ 70 và không còn con.

→ `False`

## 3. Delete

Ý tưởng: chỉ xóa ở lá, nếu lá thiếu khóa thì mượn hoặc gộp.

Bước 1: đưa về lá. Nếu khóa nằm ở nút trong thì thay nó bằng khóa kế tiếp (successor, nhỏ nhất của cây con bên phải) nằm ở lá, rồi xóa khóa ở lá.

Bước 2: xóa ở lá.

Lá là nút 3 (2 khóa): bỏ khóa, còn nút 2, xong.
Lá là nút 2 (1 khóa): bỏ khóa thì lá rỗng (underflow), cần sửa.

Bước 3: sửa underflow (nút rỗng x, cha p, anh em s):

Trường hợp	Cách xử lý
Anh em s có 2 khóa	Mượn (xoay): khóa của cha ngăn giữa x và s đi xuống x, khóa gần nhất của s đi lên cha. Xong, không lan tiếp.
Anh em s chỉ có 1 khóa	Gộp: kéo khóa của cha xuống gộp với s thành nút 3, bỏ nút rỗng. Cha mất 1 khóa nên có thể underflow, lặp lại bước 3 ở cha.

Bước 4: ở gốc. Nếu gốc bị rỗng sau khi gộp thì bỏ gốc, con duy nhất thành gốc mới. Đây là cách duy nhất cây thấp đi.

Độ phức tạp: O(log N).
# Xóa

### `delete(25)`
1. 25 nằm ở lá `[25]` (con giữa của `[20, 30]`). Xóa xong lá rỗng.
2. Hai anh em `[10]` và `[35]` đều chỉ có 1 khóa, không mượn được.
3. Gộp với anh em trái: kéo 20 từ cha xuống, được lá `[10, 20]`. Cha còn `[30]` với hai con `[10, 20]` và `[35]`.
4. Cha vẫn còn 1 khóa nên không underflow, dừng.

```
          [40]
        /      \
     [30]       [60]
     /  \       /  \
[10,20] [35]  [50] [70]
```
→ `True`, traversal: `[10, 20, 30, 35, 40, 50, 60, 70]`

### `delete(30)`
1. 30 nằm ở nút trong. Successor là 35, nằm ở lá `[35]`. Thay 30 bằng 35, nút cha thành `[35]`, rồi xóa 35 ở lá, lá rỗng.
2. Anh em trái `[10, 20]` có 2 khóa nên mượn: đẩy 20 lên cha, kéo 35 xuống lá rỗng.
3. Không còn underflow.

```
          [40]
        /      \
     [20]       [60]
     /  \       /  \
  [10] [35]   [50] [70]
```
→ `True`, traversal: `[10, 20, 35, 40, 50, 60, 70]`

### `delete(10)`
1. Xóa 10 ở lá `[10]`, lá rỗng.
2. Anh em `[35]` chỉ có 1 khóa. Gộp: kéo 20 xuống thành lá `[20, 35]`. Cha `[20]` giờ rỗng, chỉ còn 1 con, nên underflow lan lên.
3. Nút rỗng này có anh em `[60]` (1 khóa) nên không mượn được. Gộp với 40 của gốc và 60, được nút `[40, 60]` với 3 con `[20, 35]`, `[50]`, `[70]`.
4. Gốc cũ rỗng nên bị bỏ đi, cây thấp xuống 1 tầng.

```
        [40, 60]
       /    |    \
 [20, 35]  [50]  [70]
```
→ `True`

### `delete(20)`
1. 20 nằm ở lá `[20, 35]` có 2 khóa, xóa trực tiếp, lá còn `[35]`.
2. Không underflow.

```
        [40, 60]
       /    |    \
    [35]  [50]  [70]
```
→ `True`, traversal: `[35, 40, 50, 60, 70]`

## 4. Cài đặt bằng Python

### 4.1. Cấu trúc module và mối quan hệ giữa các lớp

Code được tách thành các module nhỏ, mỗi ô code bên dưới là nội dung của một file `.py` trùng tên với dòng `import` trong ô `TwoThreeTree`:

Mỗi lớp thao tác (`Search`, `Insert`, `Delete`, `InOrderTranversal`) không có `__init__` riêng, chỉ dùng thuộc tính `self.root`. `TwoThreeTree` kế thừa cả bốn lớp (đa kế thừa kiểu *mixin*) và tự khai báo `self.root = None`. Nhờ vậy mỗi thuật toán nằm gọn trong một file, dễ đọc và dễ kiểm tra riêng.

> **Lưu ý khi chạy lại:** các ô có `from Node import Node23`, `from Tree import TwoThreeTree`... cần các file `.py` tương ứng nằm cùng thư mục với notebook (nội dung mỗi file chính là nội dung ô code cùng tên bên dưới). Nếu chạy trên máy khác, hãy lưu từng ô thành file trước.

Ta đi lần lượt từ nút, cây, rồi tới các thao tác: Insert, Traversal, Search, Delete.

````python
class Node23:
    
    def __init__(self, key=None):
        self.keys = [] if key is None else [key]
        self.parent = None
        self.children = []

    def is_leaf(self):
        return len(self.children) == 0

    def is_full(self):
        return len(self.keys) == 3
````

**Giải thích `Node23`**

* **Đầu vào:** một khóa tùy chọn. Nếu có, nút khởi tạo với `keys = [key]`, nếu không thì `keys = []`.
* **Ba thuộc tính:** `keys` (khóa, luôn giữ tăng dần), `children` (các nút con), `parent` (nút cha, `None` với gốc).
* **Quy ước ghép khóa và con:** nếu nút có khóa `k0 < k1` thì `children[0]` chứa các khóa nhỏ hơn `k0`, `children[1]` chứa khóa nằm giữa `k0` và `k1`, `children[2]` chứa khóa lớn hơn `k1`. Đây chính là bảng phân hoạch khoảng giá trị ở phần 2.1.
* **Một lớp cho mọi loại nút:** nút 2 là `len(keys) == 1` (2 con), nút 3 là `len(keys) == 2` (3 con), và **nút 4 tạm thời** là `len(keys) == 3` (4 con), chỉ tồn tại thoáng qua trước khi bị tách.
* **`is_leaf()`:** nút là lá khi không có con. Cây 2-3 hợp lệ không có nút trong nào thiếu con, nên điều kiện này là đủ.
* **`is_full()`:** nút có 3 khóa, tức là đã tràn thành nút 4. Đây là điều kiện kích hoạt `_split()` trong Insert.
* **Vì sao cần `parent`:** khi tách hoặc gộp, ta phải đi ngược từ lá lên gốc. Có con trỏ cha thì không cần đệ quy từ gốc hay lưu đường đi. Cái giá là mỗi khi đổi con của một nút phải nhớ cập nhật `child.parent` (xem `_split` và `_fix_underflow`).

````python
from Node import Node23
from Insertion import Insert
from InOrderTranversal import InOrderTranversal
from Deletion import Delete
from Search import Search


class TwoThreeTree(Search, Insert, Delete, InOrderTranversal):
    def __init__(self):
        self.root = None
````

**Giải thích `TwoThreeTree`**

* Lớp này **không tự cài thuật toán nào**. Nó chỉ ghép `Search`, `Insert`, `Delete`, `InOrderTranversal` lại bằng đa kế thừa, và đặt `self.root = None` (**cây rỗng**).
* Vì cây rỗng được biểu diễn bằng `root = None`, mọi thao tác đều phải xử lý riêng trường hợp này: `_search` kiểm tra `node is None`, `insert` kiểm tra `self.root is None` để tạo nút gốc đầu tiên, `_traverse` bỏ qua `None`, và `delete` trả về `False` khi `_find_node` không thấy khóa.
* Các phần tiếp theo trình bày từng thao tác theo cùng một nhịp: ý tưởng, code, giải thích luồng chạy, rồi kết quả chạy thật.

````python
from Node import Node23

class Search:
    
    def search(self, key):
            return self._search(self.root, key)
    
    def _search(self, node, key):
            if node is None:
                return False
    
            if key in node.keys:
                return True
    
            if node.is_leaf():
                return False
    
            if key < node.keys[0]:
                return self._search(node.children[0], key)
    
            if len(node.keys) == 1 or key < node.keys[1]:
                return self._search(node.children[1], key)
    
            return self._search(node.children[2], key)
````

> Ô code ngay trên là lớp `Search` (file `Search.py`). Nội dung này **giống hệt** ô trong mục **SEARCH** ở phía dưới (ô dưới có kèm kết quả chạy). Ô này được giữ lại vì lớp `TwoThreeTree` ở trên phụ thuộc vào module `Search`; phần giải thích chi tiết nằm ở mục SEARCH.

**INSERT**


````python
from Node import Node23

class Insert:
    
    def insert(self, key):
        if self.root is None:
            self.root = Node23(key)
            return

        # Không chèn khóa trùng
        if self.search(key):
            return

        leaf = self._find_leaf(self.root, key)
        self._insert_into_leaf(leaf, key)

    def _find_leaf(self, node, key):
        if node.is_leaf():
            return node

        if key < node.keys[0]:
            return self._find_leaf(node.children[0], key)

        if len(node.keys) == 1 or key < node.keys[1]:
            return self._find_leaf(node.children[1], key)

        return self._find_leaf(node.children[2], key)

    def _insert_into_leaf(self, leaf, key):
        leaf.keys.append(key)
        leaf.keys.sort()

        if leaf.is_full():
            self._split(leaf)

    def _split(self, node):
        # node có 3 khóa: [a, b, c]
        left_key = node.keys[0]
        mid_key = node.keys[1]
        right_key = node.keys[2]

        left_node = Node23(left_key)
        right_node = Node23(right_key)

        # Nếu là nút trong, chia lại 4 cây con
        if not node.is_leaf():
            left_node.children = node.children[:2]
            right_node.children = node.children[2:]

            for child in left_node.children:
                child.parent = left_node

            for child in right_node.children:
                child.parent = right_node

        parent = node.parent

        # Tách nút gốc
        if parent is None:
            new_root = Node23(mid_key)
            new_root.children = [left_node, right_node]

            left_node.parent = new_root
            right_node.parent = new_root

            self.root = new_root
            return

        # Thay node bằng hai nút mới
        idx = parent.children.index(node)

        parent.children[idx] = left_node
        parent.children.insert(idx + 1, right_node)

        left_node.parent = parent
        right_node.parent = parent

        # Đưa khóa giữa lên cha
        parent.keys.append(mid_key)
        parent.keys.sort()

        # Sắp xếp các cây con theo thứ tự khóa
        parent.children.sort(
            key=lambda child: min(child.keys)
        )

        if parent.is_full():
            self._split(parent)
````

**Giải thích Insert**

`insert(key)` gồm bốn bước, mỗi bước là một hàm riêng:

1. **`insert`:** nếu cây rỗng thì tạo gốc `Node23(key)` và dừng. Nếu `self.search(key)` đã thấy khóa thì dừng luôn: cây này lưu **tập khóa phân biệt**, khóa trùng bị bỏ qua. (Chi phí: thêm một lần đi xuống, vẫn là $\mathcal{O}(\log N)$.)
2. **`_find_leaf(node, key)`:** đi xuống theo cùng luật rẽ nhánh với `search` (nhỏ hơn `keys[0]` đi con 0; với nút 2, hoặc nhỏ hơn `keys[1]`, đi con 1; còn lại đi con 2) cho tới khi gặp lá.
3. **`_insert_into_leaf(leaf, key)`:** thêm khóa vào lá rồi `sort()` (tối đa 3 phần tử nên rẻ). Lá là nút 2 thì thành nút 3, xong. Lá đã là nút 3 thì thành nút 4 và `is_full()` đúng, gọi `_split`.
4. **`_split(node)`:** nút có 3 khóa `[a, b, c]`:
   * Tạo `left_node` chứa `a` và `right_node` chứa `c`.
   * Nếu `node` là nút trong, 4 con được chia đôi: hai con đầu cho `left_node`, hai con sau cho `right_node`, đồng thời cập nhật `child.parent`.
   * **Nếu `node` là gốc** (`parent is None`): tạo gốc mới chứa `b` với hai con `left_node`, `right_node`. Đây là **cách duy nhất cây cao thêm**.
   * Ngược lại: thay `node` trong danh sách con của cha bằng `left_node`, chèn `right_node` ngay sau, rồi đẩy `b` vào `parent.keys`. Nếu cha đầy thì gọi lại `_split(parent)`, đó là phần "lan truyền lên trên".

**TRANVERSAL**


````python
from Node import Node23

class InOrderTranversal:
    
    def in_order(self):
        result = []
        self._traverse(self.root, result)
        return result

    def _traverse(self, node, result):
        if node is None:
            return

        if node.is_leaf():
            result.extend(node.keys)
            return

        for i, key in enumerate(node.keys):
            self._traverse(node.children[i], result)
            result.append(key)

        self._traverse(node.children[len(node.keys)], result)
````

**Giải thích `in_order` và kết quả**

* **Ý tưởng:** duyệt trung thứ tự (in-order) của cây 2-3 giống BST nhưng mỗi nút có thể có 2 khóa: duyệt con 0, ghi `keys[0]`, duyệt con 1, ghi `keys[1]` (nếu có), rồi duyệt con cuối.
* **Trong code:** lá thì ghi toàn bộ khóa. Nút trong thì vòng `for i, key in enumerate(node.keys)` gọi `_traverse(node.children[i])` rồi `result.append(key)`; sau vòng lặp gọi nốt `_traverse(node.children[len(node.keys)])` cho con cuối.
* **Độ phức tạp:** mỗi nút được thăm đúng một lần, $\mathcal{O}(N)$.
* **Kết quả:** `[10, 20, 25, 30, 35, 40, 50, 60, 70]` tăng dần và đủ 9 khóa. Điều này cho thấy **tính chất thứ tự tìm kiếm** được giữ đúng sau chuỗi chèn có tách nút. Lưu ý đây mới chỉ là bằng chứng cho tính chất thứ tự. Tính chất "mọi lá cùng độ sâu" được kiểm tra riêng ở mục 4.3.

**Giải thích Search và kết quả**

`_search(node, key)` là hàm đệ quy, mỗi lần gọi xuống đúng một tầng:

1. `node is None` (cây rỗng) thì trả `False`.
2. `key in node.keys` thì trả `True`. Kiểm tra bằng nhau đặt **trước** các phép so sánh rẽ nhánh, nên phía sau chỉ cần so sánh `<`.
3. Nếu là lá mà chưa thấy thì trả `False` (không còn nhánh để đi).
4. `key < node.keys[0]` thì đi con 0.
5. `len(node.keys) == 1 or key < node.keys[1]` thì đi con 1. Với nút 2, khóa lớn hơn `keys[0]` nên đi con "phải", chính là `children[1]`. Với nút 3, khóa nằm giữa hai khóa nên đi con giữa.
6. Còn lại đi con 2.

* `search(99)`: gốc `[40]` rẽ phải, nút `[60]` rẽ phải, tới lá `[70]` khác 99 và là lá. Output `False`.

Số bước tối đa bằng chiều cao cây nên $\mathcal{O}(\log N)$. Trường hợp tốt nhất là $\mathcal{O}(1)$ khi khóa nằm ngay ở gốc.

**DELETE**


````python
from Node import Node23

class Delete: 
    
    def delete(self, key):
        node = self._find_node(self.root, key)

        if node is None:
            return False

        # Nếu là nút trong, thay bằng successor
        if not node.is_leaf():
            idx = node.keys.index(key)
            successor_leaf, successor_key = self._get_successor(
                node, idx
            )

            node.keys[idx] = successor_key
            node = successor_leaf
            key = successor_key

        node.keys.remove(key)

        if len(node.keys) == 0:
            self._fix_underflow(node)

        return True

    def _find_node(self, node, key):
        if node is None:
            return None

        if key in node.keys:
            return node

        if node.is_leaf():
            return None

        if key < node.keys[0]:
            return self._find_node(node.children[0], key)

        if len(node.keys) == 1 or key < node.keys[1]:
            return self._find_node(node.children[1], key)

        return self._find_node(node.children[2], key)

    def _get_successor(self, node, key_idx):
        curr = node.children[key_idx + 1]

        while not curr.is_leaf():
            curr = curr.children[0]

        return curr, curr.keys[0]

    def _fix_underflow(self, node):
        # Trường hợp nút gốc bị rỗng
        if node == self.root:
            if len(node.keys) == 0:
                if node.children:
                    self.root = node.children[0]
                    self.root.parent = None
                else:
                    self.root = None
            return

        parent = node.parent
        idx = parent.children.index(node)

        left_sib = (
            parent.children[idx - 1] if idx > 0 else None
        )
        right_sib = (
            parent.children[idx + 1]
            if idx + 1 < len(parent.children)
            else None
        )

        # 1. Mượn khóa từ anh em bên trái
        if left_sib is not None and len(left_sib.keys) == 2:
            node.keys.insert(0, parent.keys[idx - 1])
            parent.keys[idx - 1] = left_sib.keys.pop()

            if left_sib.children:
                child = left_sib.children.pop()
                node.children.insert(0, child)
                child.parent = node

            return

        # 2. Mượn khóa từ anh em bên phải
        if right_sib is not None and len(right_sib.keys) == 2:
            node.keys.append(parent.keys[idx])
            parent.keys[idx] = right_sib.keys.pop(0)

            if right_sib.children:
                child = right_sib.children.pop(0)
                node.children.append(child)
                child.parent = node

            return

        # 3. Không mượn được: gộp với anh em bên trái
        if left_sib is not None:
            parent_key = parent.keys.pop(idx - 1)
            parent.children.pop(idx)

            left_sib.keys.append(parent_key)
            left_sib.keys.extend(node.keys)

            for child in node.children:
                child.parent = left_sib

            left_sib.children.extend(node.children)

        # 4. Không có anh em trái: gộp với anh em bên phải
        elif right_sib is not None:
            parent_key = parent.keys.pop(idx)
            parent.children.pop(idx + 1)

            node.keys.append(parent_key)
            node.keys.extend(right_sib.keys)

            for child in right_sib.children:
                child.parent = node

            node.children.extend(right_sib.children)

        # Nếu cha bị rỗng, xử lý tiếp từ cha
        if len(parent.keys) == 0:
            self._fix_underflow(parent)
````

**Giải thích Delete**

Delete là thao tác khó nhất vì xóa có thể làm một nút **rỗng** (underflow). Toàn bộ thuật toán xoay quanh việc *xóa ở lá* rồi *vá chỗ rỗng*.

**`delete(key)`**

1. `_find_node` tìm nút chứa `key` (cùng luật rẽ nhánh với search). Không thấy thì trả `False`.
2. Nếu nút chứa `key` là **nút trong**, `_get_successor` đi sang con ngay sau khóa đó rồi đi trái nhất tới lá, lấy khóa nhỏ nhất ở lá đó (successor). Code ghi đè `key` bằng successor rồi chuyển việc xóa xuống lá. Thay bằng successor giữ nguyên thứ tự vì đó là khóa nhỏ nhất lớn hơn `key`.
3. `node.keys.remove(key)` xóa ở lá. Lá còn khóa thì xong. Lá rỗng thì gọi `_fix_underflow(node)`.

**`_fix_underflow(node)`**: gọi `x` là nút rỗng, `p` là cha, `s` là anh em. Khóa của cha ngăn giữa `children[i]` và `children[i+1]` là `parent.keys[i]`, nên khóa ngăn với anh em trái là `parent.keys[idx - 1]`, với anh em phải là `parent.keys[idx]`.

| Thứ tự thử | Điều kiện | Xử lý | Hệ quả |
| :---: | :--- | :--- | :--- |
| 0 | `x` là gốc | Con duy nhất (nếu có) thành gốc mới, hoặc cây thành rỗng | **Cây thấp đi 1 tầng**, dừng |
| 1 | Anh em trái có 2 khóa | **Mượn trái:** khóa cha đi xuống đầu `x`, khóa lớn nhất của anh em trái đi lên cha; nếu là nút trong, con cuối của anh em trái chuyển sang đầu `x` | Dừng |
| 2 | Anh em phải có 2 khóa | **Mượn phải:** đối xứng với trên | Dừng |
| 3 | Có anh em trái (chỉ 1 khóa) | **Gộp trái:** kéo khóa cha xuống gộp `x` vào anh em trái, chuyển nốt con của `x` sang | Cha mất 1 khóa |
| 4 | Chỉ có anh em phải (chỉ 1 khóa) | **Gộp phải:** đối xứng với trên | Cha mất 1 khóa |
| 5 | Cha hết khóa sau khi gộp | Gọi lại `_fix_underflow(parent)` | Lan lên một tầng |

Hai điểm cần nhớ: code **ưu tiên mượn trước, gộp sau** (mượn dừng ngay, gộp thì có thể lan tiếp), và **ưu tiên anh em bên trái**. Chọn bên nào thì cây vẫn đúng, chỉ khác hình dạng cây kết quả.

Sau khi gộp, nút trong từng rỗng vẫn được xử lý đúng: nó có đúng 1 con, nên khi gộp, khóa cha cộng với con đó làm anh em có thêm khóa và thêm con, giữ nguyên quan hệ "số con = số khóa + 1".

### 4.2. Giải thích các phần quan trọng trong Code Python

1. **Cách biểu diễn nút (`class Node23`):**
   * Thay vì tách riêng `class TwoNode` và `class ThreeNode` (gây phức tạp khi phải ép kiểu qua lại), ta dùng danh sách động `self.keys` và `self.children`.
   * Điều này cho phép một nút **tạm thời giữ 3 khóa và 4 con** (`is_full() == True`) ngay trước khi bị tách, khớp hoàn toàn với khái niệm **4-node tạm thời** trong lý thuyết. Nút còn có thêm con trỏ `parent` để đi ngược lên khi tách hoặc gộp.
2. **Cơ chế đẩy khóa ngược lên trên trong `_split()`:**
   * Việc chèn không đệ quy từ gốc xuống rồi quay lui. Thay vào đó `_find_leaf()` tìm lá, `_insert_into_leaf()` nhét khóa vào lá, và nếu lá đầy thì `_split()` đi ngược lên bằng con trỏ `parent`.
   * Hai dòng `parent.keys.append(mid_key)` + `parent.keys.sort()` đẩy khóa giữa lên cha. Hai dòng `parent.children[idx] = left_node` và `parent.children.insert(idx + 1, right_node)` thay nút cũ bằng hai nút mới. Vì mỗi nút chỉ có tối đa 4 con nên các thao tác danh sách này tốn $\mathcal{O}(1)$.
   * Nếu cha cũng đầy thì `_split(parent)` được gọi lại, nên việc tách có thể lan tới gốc.
3. **Độ phức tạp thời gian & không gian:**
   * **Tìm kiếm (`search`) & Chèn (`insert`):** Đều tỉ lệ thuận với chiều cao cây $H \le \log_2(N+1) - 1$, do đó đảm bảo độ phức tạp **$\mathcal{O}(\log N)$ trong trường hợp xấu nhất**. Lưu ý `insert` gọi `search` một lần để loại khóa trùng nên đi xuống hai lần, vẫn là $\mathcal{O}(\log N)$.
   * **Xóa (`delete`):** đi xuống lá một lần, sau đó sửa underflow có thể lan ngược lên tối đa $H$ tầng, mỗi tầng $\mathcal{O}(1)$, tổng $\mathcal{O}(\log N)$.
   * **Ý nghĩa thực tiễn:** Cây 2-3 chính là tiền đề trực tiếp để phát triển **Cây Đỏ-Đen (Red-Black Tree)** (dùng trong `std::map` của C++, `TreeMap` của Java) và **B-Tree** (dùng trong cơ sở dữ liệu SQL và hệ thống file).

### 4.3. Kiểm thử: in hình dạng cây và kiểm tra các bất biến

Các ví dụ ở trên chỉ chứng minh code đúng **trên vài trường hợp**. Để tăng độ tin cậy, ta kiểm tra trực tiếp các **bất biến** (tính chất luôn phải đúng) của cây 2-3 sau mỗi thao tác:

1. Mỗi nút có 1 hoặc 2 khóa, khóa trong nút tăng dần.
2. Nút trong có đúng `số khóa + 1` con; lá không có con.
3. **Thứ tự tìm kiếm:** mọi khóa trong con thứ `i` nằm trong khoảng giữa hai khóa kề của nút cha.
4. **Cân bằng tuyệt đối:** mọi lá cùng một độ sâu.
5. Con trỏ `parent` của mỗi nút trỏ đúng nút cha.

Ô code dưới định nghĩa hai hàm: `print_levels` (in cây theo từng tầng để nhìn hình dạng) và `check_invariants` (kiểm tra năm bất biến trên, báo `AssertionError` nếu vi phạm, trả về độ sâu chung của các lá). Sau đó dựng lại ví dụ ở phần 3 trên một cây mới `demo` và in hình dạng sau từng bước quan trọng.

````python
def print_levels(t):
    """In cây theo từng tầng, mỗi nút viết dạng [k1|k2]."""
    if t.root is None:
        print("  (cây rỗng)")
        return
    level = [t.root]
    d = 0
    while level:
        print(f"  tầng {d}: " + "  ".join("[" + "|".join(map(str, n.keys)) + "]" for n in level))
        level = [c for n in level for c in n.children]
        d += 1


def check_invariants(t):
    """Kiểm tra các bất biến của cây 2-3. Trả về độ sâu chung của các lá (None nếu cây rỗng)."""
    leaf_depths = set()

    def rec(n, lo, hi, depth, parent):
        assert n.parent is parent, "con trỏ parent sai"
        assert 1 <= len(n.keys) <= 2, f"số khóa không hợp lệ: {n.keys}"
        assert n.keys == sorted(n.keys), "khóa trong nút không tăng dần"
        for k in n.keys:
            assert (lo is None or k > lo) and (hi is None or k < hi), "vi phạm thứ tự tìm kiếm"
        if n.is_leaf():
            leaf_depths.add(depth)
            return
        assert len(n.children) == len(n.keys) + 1, "số con != số khóa + 1"
        bounds = [lo] + n.keys + [hi]
        for i, c in enumerate(n.children):
            rec(c, bounds[i], bounds[i + 1], depth + 1, n)

    if t.root is not None:
        rec(t.root, None, None, 0, None)
    assert len(leaf_depths) <= 1, f"các lá không cùng độ sâu: {leaf_depths}"
    return next(iter(leaf_depths)) if leaf_depths else None


# Dựng lại ví dụ ở phần 3 trên một cây mới
demo = TwoThreeTree()
for v in [30, 20, 10, 50, 40, 70, 60, 25, 35]:
    demo.insert(v)
print("Sau khi chèn xong (độ sâu lá =", check_invariants(demo), "):")
print_levels(demo)

for k in [25, 30, 10, 20]:
    demo.delete(k)
    print(f"\nSau delete({k}) (độ sâu lá =", check_invariants(demo), "):")
    print_levels(demo)
````

**Đọc kết quả**

* **Sau khi chèn xong:** gốc `[40]`, tầng 1 gồm `[20|30]` và `[60]`, tầng 2 gồm năm lá `[10] [25] [35] [50] [70]`, đúng sơ đồ ở phần 3.1. Độ sâu của lá là 2 và `check_invariants` không báo lỗi.
* **Sau `delete(25)`:** nút `[20|30]` trở thành `[30]` với hai con `[10|20]` và `[35]`, khớp phần 3.3.
* **Sau `delete(30)`:** gốc nhánh trái thành `[20]` với hai lá `[10]` và `[35]`, khớp phần 3.3.
* **Sau `delete(10)`:** gộp dây chuyền, gốc thành `[40|60]` và **độ sâu lá giảm từ 2 xuống 1**. Đây là lần duy nhất chiều cao giảm trong ví dụ, đúng nguyên lý "chỉ gốc mới làm đổi chiều cao".
* **Sau `delete(20)`:** lá `[20|35]` chỉ còn `[35]`, không cần sửa gì thêm.

Hình dạng cây thật khớp từng bước với trace tay. Tiếp theo ta kiểm tra trên **dữ liệu ngẫu nhiên lớn** thay vì một ví dụ nhỏ.

````python
import random

random.seed(1)                       # cố định seed để kết quả lặp lại được
so_lan, so_thao_tac = 300, 300
for _ in range(so_lan):
    t = TwoThreeTree()
    tap_tham_chieu = set()           # dùng set của Python làm "đáp án chuẩn"
    for _ in range(so_thao_tac):
        k = random.randint(0, 60)    # miền nhỏ để có nhiều lần chèn trùng và xóa khóa không tồn tại
        if random.random() < 0.55:
            t.insert(k)
            tap_tham_chieu.add(k)
        else:
            assert t.delete(k) == (k in tap_tham_chieu)
            tap_tham_chieu.discard(k)
        check_invariants(t)                              # các bất biến vẫn đúng
        assert t.in_order() == sorted(tap_tham_chieu)    # nội dung khớp tập chuẩn
print(f"Ngẫu nhiên: {so_lan} cây x {so_thao_tac} thao tác chèn/xóa, bất biến đúng ở mọi bước.")

# Dữ liệu tăng dần: kịch bản xấu nhất của BST thường
n = 1023
t = TwoThreeTree()
for i in range(1, n + 1):
    t.insert(i)
print(f"Chèn tăng dần 1..{n}: độ sâu mọi lá = {check_invariants(t)}  (BST thường sẽ sâu {n - 1})")

# Xóa hết từng khóa
for i in range(1, n + 1):
    assert t.delete(i)
print("Xóa hết:", "root =", t.root, ", in_order =", t.in_order())
````

**Phân tích kết quả**

* **Kiểm thử ngẫu nhiên:** 300 cây, mỗi cây 300 thao tác trộn chèn và xóa trên miền khóa nhỏ (0 đến 60). Miền nhỏ cố ý tạo nhiều tình huống khó: chèn khóa trùng (bị bỏ qua), xóa khóa không tồn tại (trả `False`), và tách/gộp dây chuyền. Sau **mỗi thao tác** ta kiểm tra bất biến và so sánh với `set` của Python. Không có lỗi nào xuất hiện trong lần chạy này.
* **Dữ liệu tăng dần 1 đến 1023:** BST thường sẽ thoái hóa thành danh sách liên kết sâu 1022 mức. Cây 2-3 cho **mọi lá cùng độ sâu 9**, đúng bằng $\log_2(1024) - 1$, tức là cây nhị phân đầy hoàn hảo (toàn nút 2). Đây là minh chứng trực tiếp cho cận $H \le \log_2(N+1) - 1$ ở phần 2.1.
* **Xóa hết:** cây về `root = None`, duyệt ra danh sách rỗng, nghĩa là các trường hợp biên của delete (xóa khóa cuối cùng, gốc rỗng) chạy đúng.

**Giới hạn của kiểm thử này:** nó là kiểm thử ngẫu nhiên với seed cố định trên khóa nguyên, không phải chứng minh hình thức. Nó chưa kiểm tra hiệu năng, kiểu khóa khác số nguyên, hay dữ liệu cực lớn.

### 4.4. Độ phức tạp và các trường hợp biên

**Cận chiều cao (cơ sở của mọi phân tích).** Gọi $N$ là số khóa, $H$ là chiều cao tính theo số cạnh. Mọi lá cùng độ sâu $H$ và mọi nút trong có ít nhất 2 con, nên số khóa tối thiểu là $\sum_{i=0}^{H} 2^i = 2^{H+1} - 1$, suy ra $H \le \log_2(N+1) - 1$. Ngược lại, mỗi nút có tối đa 3 con và 2 khóa nên $N \le 3^{H+1} - 1$, suy ra $H \ge \log_3(N+1) - 1$. Hai cận chỉ phụ thuộc hai bất biến (thứ tự và lá cùng độ sâu), mà insert và delete đều bảo toàn, nên cây **không bao giờ thoái hóa**, bất kể thứ tự dữ liệu.

**Độ phức tạp thời gian**

| Thao tác | Tốt nhất | Trung bình | Xấu nhất | Ghi chú |
| :--- | :---: | :---: | :---: | :--- |
| `search` | $\mathcal{O}(1)$ (khóa ở gốc) | $\Theta(\log N)$ | $\Theta(\log N)$ | Đi một đường từ gốc xuống lá |
| `insert` | $\Theta(\log N)$ | $\Theta(\log N)$ | $\Theta(\log N)$ | Xuống lá; xấu nhất là tách dây chuyền tới gốc, mỗi tầng $\mathcal{O}(1)$ |
| `delete` | $\Theta(\log N)$ | $\Theta(\log N)$ | $\Theta(\log N)$ | Xuống lá; xấu nhất là gộp dây chuyền tới gốc |
| `in_order` | $\Theta(N)$ | $\Theta(N)$ | $\Theta(N)$ | Thăm mỗi nút một lần |

Về trường hợp trung bình: cây thực tế là hỗn hợp nút 2 và nút 3 nên chiều cao nằm giữa hai cận trên, và không có công thức đóng gọn cho trung bình. Tuy vậy mọi trường hợp đều bị chặn trên bởi $\log_2(N+1)$ nên kết luận $\Theta(\log N)$ vẫn đúng.

**Độ phức tạp không gian:** $\mathcal{O}(N)$. Số nút nằm giữa khoảng $N/2$ (toàn nút 3 đầy) và $N$ (toàn nút 2). Các hàm đệ quy (`_search`, `_find_leaf`, `_split`, `_fix_underflow`) chỉ đệ quy tối đa bằng chiều cao nên cần $\mathcal{O}(\log N)$ bộ nhớ ngăn xếp, không lo vượt giới hạn đệ quy của Python.

**Các trường hợp biên đã được kiểm tra bằng code ở mục 4.3**

| Tình huống | Hành vi của code |
| :--- | :--- |
| Cây rỗng | `search` trả `False`, `delete` trả `False`, `in_order` trả `[]` |
| Chèn khóa đã có | Bị bỏ qua (nhờ `self.search` trong `insert`), cây không đổi |
| Xóa khóa không tồn tại | Trả `False`, cây không đổi |
| Xóa khóa cuối cùng | Gốc về `None`, cây rỗng |
| Chèn dữ liệu tăng dần | Cây vẫn cân bằng tuyệt đối (độ sâu lá 9 với 1023 khóa) |

**Những chỗ thuật toán hoặc code chưa hiệu quả (nên chủ động nêu khi thuyết trình)**

* **`insert` đi xuống hai lần:** một lần ở `search` để loại trùng, một lần ở `_find_leaf`. Vẫn $\mathcal{O}(\log N)$ nhưng gấp đôi hằng số. Có thể gộp thành một lần đi xuống.
* **`_split` tạo nút mới** thay vì tái sử dụng nút cũ, và dòng `children.sort(...)` thừa như đã phân tích. Cả hai chỉ ảnh hưởng hằng số.
* **Chèn và xóa xen kẽ:** cây 2-3 không có "khoảng đệm" giữa ngưỡng tách (3 khóa) và ngưỡng gộp (0 khóa). Vì vậy một chuỗi thao tác xen kẽ có thể làm *mỗi* thao tác phải cấu trúc lại tới $\Theta(\log N)$ tầng. Cây 2-3-4 và Red-Black cổ điển khắc phục được điều này, là một lý do chúng phổ biến hơn trong thực tế.
* **Chỉ lưu khóa**, chưa lưu giá trị đi kèm, nên mới là *tập có thứ tự*, chưa phải *từ điển*.
* **Chi phí bộ nhớ vật lý:** mỗi nút phải cấp phát chỗ cho cả trường hợp 2 khóa và 3 con, nên nút 2 lãng phí chỗ. Trong một mô hình cấu trúc C với khóa 64-bit, ước lượng khoảng 48 byte mỗi nút, nút 2 lãng phí gần một nửa. Con số này là ước lượng lý thuyết, **chưa đo trên code Python** (Python có thêm chi phí đối tượng và danh sách).
* **Khả năng khai thác bộ nhớ đệm (cache) hạn chế:** vẫn là duyệt con trỏ, chỉ giảm được hệ số hằng. Muốn khai thác phân cấp bộ nhớ cần B-Tree bậc cao, mỗi nút bằng một trang đĩa.

Những hạn chế về số nhánh xử lý (nút 4 tạm, mượn trái/phải, gộp trái/phải) cũng là lý do thực tế người ta mã hóa cây 2-3 thành Red-Black Tree, nội dung của phần tiếp theo.

## 5. Cài đặt thực tế: mã hóa cây 2-3 thành LLRB

### 5.1. Vì sao hiếm khi code cây 2-3 trực tiếp?

Code ở phần 4 phải xử lý quá nhiều nhánh: nút 2, nút 3 và nút 4 tạm; mượn trái, mượn phải; gộp trái, gộp phải, mỗi ca lại có biến thể đối xứng. Điều đó dễ sai và khiến CPU khó dự đoán rẽ nhánh (*branch misprediction*). Thực tế người ta **mã hóa cây 2-3 thành cây nhị phân** có thêm 1 bit màu cho mỗi liên kết: **Left-Leaning Red-Black Tree (LLRB)**.

| Cây 2-3 | Trong LLRB |
| :--- | :--- |
| Nút 2 `[a]` | Một nút **đen** `a` |
| Nút 3 `[a\|b]` | Nút **đen** `b` có con **trái đỏ** `a` (liên kết đỏ luôn nghiêng trái) |
| Nút 4 tạm `[a\|b\|c]` | Nút **đen** `b` có **hai con đỏ** `a` và `c` |

```
Nút 3 [a|b]  <->    b (đen)            Nút 4 [a|b|c]  <->     b (đen)
                   /                                         /     \
                a (đỏ)                                    a (đỏ)   c (đỏ)
```

Liên kết đỏ "dán" hai nút nhị phân lại thành một nút 3 hoặc nút 4. Mọi đường từ gốc tới liên kết rỗng có **cùng số liên kết đen**, tương ứng đúng với bất biến "mọi lá cùng độ sâu" của cây 2-3.

### 5.2. Ba phép biến đổi thay cho toàn bộ việc tách nút

Việc tách nút dây chuyền từ lá lên gốc rút gọn thành ba phép, được áp dụng trên đường **quay lui** sau khi chèn:

| Phép | Khi nào dùng | Ý nghĩa theo cây 2-3 |
| :--- | :--- | :--- |
| `rotate_left` | Liên kết đỏ nghiêng **phải** | Chuẩn hóa nút 3 về dạng nghiêng trái |
| `rotate_right` | **Hai** liên kết đỏ liên tiếp bên trái | Biến nút 4 dạng "chuỗi" thành nút 4 cân đối |
| `flip_colors` | Nút có **hai con đỏ** (nút 4) | **Tách nút 4:** hai con thành đen, nút cha thành đỏ, tức là đẩy khóa giữa lên tầng trên |

Cần phân biệt: **LLRB tương đương đúng cây 2-3**, còn Red-Black cổ điển cho phép nghiêng cả hai phía nên tương đương **cây 2-3-4**. `TreeMap` của Java và `std::map` của C++ dùng Red-Black cổ điển chứ không phải LLRB; LLRB chủ yếu dùng để giảng dạy và cài đặt nhanh.

## 6. Ứng dụng thực tế và bài tập

### 6.1. Ứng dụng trong hệ thống

Cây 2-3 là nền tảng lý thuyết cho hai họ cấu trúc dùng rất rộng:

| Họ cấu trúc | Quan hệ với cây 2-3 | Ví dụ sử dụng |
| :--- | :--- | :--- |
| **B-Tree / B+Tree** | Tổng quát hóa cây 2-3 với bậc rất lớn, mỗi nút bằng một trang đĩa | Index của MySQL (InnoDB), PostgreSQL, SQLite; metadata của hệ thống file NTFS, Btrfs, XFS, APFS |
| **Red-Black Tree** cổ điển | Tương đương cây 2-3-4 | `TreeMap`/`TreeSet` của Java, `std::map`/`std::set` của C++, nhân Linux |
| **LLRB** | Tương đương đúng cây 2-3 | Chủ yếu để giảng dạy, cài đặt nhanh một cây cân bằng |

Cây 2-3 chính là B-Tree bậc 3. Với dữ liệu nằm trong RAM, người ta dùng Red-Black; với dữ liệu trên đĩa hoặc trong cơ sở dữ liệu, người ta dùng B-Tree.

### 6.2. Bài tập LeetCode 315: Count of Smaller Numbers After Self

**Đề:** với mỗi phần tử `nums[i]`, đếm số phần tử **nhỏ hơn** nó nằm **bên phải** nó. Ví dụ `nums = [5, 2, 6, 1]` cho kết quả `[2, 1, 1, 0]`.

**Tư duy:** duyệt từ **phải sang trái** và giữ một tập động các phần tử đã thấy. Với mỗi phần tử `x`, hỏi "trong tập có bao nhiêu khóa nhỏ hơn `x`" (truy vấn **rank**), rồi chèn `x` vào tập. Cái ta cần là một cấu trúc vừa **chèn** vừa **đếm số khóa nhỏ hơn** đều nhanh:

| Cấu trúc | Tìm vị trí | Chèn | Tổng |
| :--- | :---: | :---: | :---: |
| Mảng đã sắp xếp | $\mathcal{O}(\log n)$ | $\mathcal{O}(n)$ (phải dịch phần tử) | $\mathcal{O}(n^2)$ |
| Danh sách liên kết | $\mathcal{O}(n)$ | $\mathcal{O}(1)$ | $\mathcal{O}(n^2)$ |
| **Cây cân bằng, mỗi nút lưu `size`** | $\mathcal{O}(\log n)$ | $\mathcal{O}(\log n)$ | $\mathcal{O}(n \log n)$ |

**Chìa khóa:** mỗi nút lưu thêm `size` (tổng số phần tử trong cây con) và `cnt` (số lần khóa xuất hiện, để xử lý trùng). Có `size`, `rank(x)` chỉ cần một lần đi từ gốc xuống: mỗi lần rẽ phải thì cộng thêm "toàn bộ cây con trái + chính nút đó". `size` chỉ cần cập nhật trong `insert` và trong các phép xoay, vì xoay chỉ thay đổi cấu trúc cục bộ. Vì `TreeMap` của Java không có hàm `rank()` nên ta tự cài một cây tăng cường; ở đây dùng LLRB vì gọn nhất (ba phép biến đổi ở mục 5.2).

````python
RED, BLACK = True, False

class Node:
    __slots__ = ("key", "cnt", "size", "left", "right", "color")
    def __init__(self, key, color):
        self.key, self.cnt, self.size, self.color = key, 1, 1, color
        self.left = self.right = None

def size(h):   return h.size if h else 0
def is_red(h): return h is not None and h.color == RED
def update(h): h.size = size(h.left) + size(h.right) + h.cnt

def rotate_left(h):             # Chuẩn hóa nút 3 nghiêng phải -> nghiêng trái
    x = h.right
    h.right, x.left = x.left, h
    x.color, h.color = h.color, RED
    x.size = h.size
    update(h)
    return x

def rotate_right(h):            # Cân đối nút 4 lệch trái
    x = h.left
    h.left, x.right = x.right, h
    x.color, h.color = h.color, RED
    x.size = h.size
    update(h)
    return x

def flip_colors(h):             # Tách nút 4: đẩy trung vị h lên cha
    h.color = RED
    h.left.color = h.right.color = BLACK

def insert(h, key):
    if h is None: return Node(key, RED)
    if   key < h.key: h.left  = insert(h.left,  key)
    elif key > h.key: h.right = insert(h.right, key)
    else:             h.cnt += 1
    
    update(h)                   # Cập nhật size TRƯỚC khi xoay
    if is_red(h.right) and not is_red(h.left):  h = rotate_left(h)
    if is_red(h.left) and is_red(h.left.left):  h = rotate_right(h)
    if is_red(h.left) and is_red(h.right):      flip_colors(h)
    return h

def rank(h, key):               # Đếm số khóa < key
    r = 0
    while h:
        if   key < h.key: h = h.left
        elif key > h.key: r += size(h.left) + h.cnt; h = h.right
        else:             return r + size(h.left)
    return r

def countSmaller(nums):
    root, res = None, [0] * len(nums)
    for i in range(len(nums) - 1, -1, -1):
        res[i] = rank(root, nums[i])
        root = insert(root, nums[i])
        root.color = BLACK
    return res
````

**Giải thích code LLRB**

**Dữ liệu:** `Node` có `key`, `cnt` (số lần khóa xuất hiện), `size` (tổng số phần tử trong cây con, **tính cả `cnt`**), `left`, `right` và `color` (`RED = True`, `BLACK = False`). Nút mới luôn **đỏ** (tương ứng "nhét khóa vào lá": nút 2 thành nút 3, nút 3 thành nút 4 tạm). `__slots__` chỉ để tiết kiệm bộ nhớ.

**Hàm phụ:** `size(h)` trả 0 cho `None`; `is_red(h)` coi `None` là đen; `update(h)` tính lại `size` theo công thức `size(left) + size(right) + cnt`.

**`rotate_left` / `rotate_right`:** xoay quanh liên kết đỏ. Nút `x` (con được nâng lên) lấy màu của `h` cũ, còn `h` trở thành đỏ. Về `size`: `x` thay chỗ `h` làm gốc của cây con nên tổng số phần tử **không đổi**, `x.size = h.size`; chỉ `h` (nay là con của `x`) cần `update(h)`.

**`flip_colors`:** hai con thành đen, `h` thành đỏ. Chính là **tách nút 4**: khóa giữa `h` được đẩy lên tầng trên (liên kết đỏ lên cha), hai khóa bên cạnh thành hai nút 2 độc lập.

**`insert(h, key)`:**
1. `h is None` thì tạo nút đỏ mới.
2. So sánh để đi trái, đi phải, hoặc nếu **bằng** thì chỉ tăng `cnt` (không tạo nút mới).
3. `update(h)` **trước khi xoay**, vì cây con vừa thay đổi.
4. Ba dòng `if` theo đúng thứ tự: (a) liên kết đỏ nghiêng phải thì `rotate_left`; (b) hai liên kết đỏ liên tiếp bên trái thì `rotate_right`; (c) hai con cùng đỏ thì `flip_colors`. Mỗi dòng xử lý một dạng nút 3 hoặc nút 4, và vì chạy trên đường quay lui nên việc "đẩy lên" tự lan từ lá tới gốc, đúng bản chất của cây 2-3.

**`rank(h, key)`:** vòng lặp đi từ gốc. Rẽ trái thì không cộng gì. Rẽ phải thì cộng `size(h.left) + h.cnt` (toàn bộ cây con trái cùng chính nút `h`, đều nhỏ hơn `key`). Gặp đúng `key` thì cộng `size(h.left)` và dừng, tức là **không tính các phần tử bằng `key`**, đúng yêu cầu "nhỏ hơn" của đề.

**`countSmaller(nums)`:** duyệt `i` từ cuối về đầu; trước khi chèn `nums[i]` thì gọi `rank`, kết quả chính là số phần tử nhỏ hơn nằm bên phải. Sau mỗi lần chèn đặt `root.color = BLACK`: `flip_colors` có thể làm gốc đỏ; tô đen lại tương ứng với việc "gốc bị tách thì tạo gốc mới, cây cao thêm 1 tầng".

**Đối chiếu với trace thủ công** trên `[5, 2, 6, 1]` (duyệt từ phải sang trái):

| `i` | `nums[i]` | Tập đã có | `rank` (số khóa < x) |
| :---: | :---: | :--- | :---: |
| 3 | 1 | rỗng | 0 |
| 2 | 6 | {1} | 1 |
| 1 | 2 | {1, 6} | 1 |
| 0 | 5 | {1, 2, 6} | 2 |

Ghép lại theo vị trí được `[2, 1, 1, 0]`.

````python
# Ví dụ của đề và các trường hợp biên
print(countSmaller([5, 2, 6, 1]))   # ví dụ của đề
print(countSmaller([-1]))           # một phần tử
print(countSmaller([-1, -1]))       # khóa trùng: không tính phần tử bằng nhau

# Đối chiếu với cách làm vét cạn O(n^2) trên nhiều mảng ngẫu nhiên (có số âm, có trùng)
import random

def brute(nums):
    return [sum(1 for y in nums[i + 1:] if y < nums[i]) for i in range(len(nums))]

random.seed(2)
for _ in range(500):
    a = [random.randint(-20, 20) for _ in range(random.randint(1, 40))]
    assert countSmaller(a) == brute(a)
print("500 mảng ngẫu nhiên: LLRB khớp vét cạn.")

# Kiểm tra cây thật sự cân bằng: chèn 10000 khóa tăng dần
import math

def height(h):
    return 0 if h is None else 1 + max(height(h.left), height(h.right))

root = None
for i in range(10000):
    root = insert(root, i)
    root.color = BLACK
print("Chiều cao (số tầng) khi chèn 10000 khóa tăng dần:", height(root),
      "| cận lý thuyết 2*log2(n+1) =", round(2 * math.log2(10001), 1))
````

**Phân tích kết quả**

* `countSmaller([5, 2, 6, 1])` ra `[2, 1, 1, 0]`, đúng đáp án mẫu của đề và khớp bảng trace ở trên.
* `[-1, -1]` ra `[0, 0]`: với phần tử đầu, trong tập đã có một khóa `-1` bằng nó, và `rank` **không** đếm phần tử bằng nhau, nên kết quả là 0. Đó là nhờ nhánh "gặp đúng khóa" trả về `r + size(h.left)`.
* Đối chiếu với cách vét cạn $\mathcal{O}(n^2)$ trên 500 mảng ngẫu nhiên đều khớp. Đây là kiểm thử **tính đúng của đáp án**, chưa phải kiểm thử hiệu năng.
* Phép kiểm tra cuối chèn 10000 khóa **đã sắp xếp** (kịch bản làm BST thường thoái hóa thành chiều cao 10000). LLRB cho chiều cao thấp hơn nhiều, nằm dưới cận lý thuyết $2\log_2(n+1)$ của LLRB. Số liệu cụ thể xem ở output ở trên.
* **Độ phức tạp:** mỗi `rank` và `insert` tốn $\mathcal{O}(\log n)$, lặp $n$ lần nên thời gian $\Theta(n \log n)$, không gian $\Theta(n)$ cho cây.

> **Lưu ý:** ô code LLRB định nghĩa các tên toàn cục `Node`, `insert`, `size`... Chúng không xung đột với các lớp của phần 4 (`Node23`, `Insert`...) nên có thể chạy trong cùng notebook.

### 6.3. Phương pháp khác (so sánh mở rộng)

Dùng cây cân bằng cho bài 315 là chưa tối ưu. Vì mảng đầu vào **biết trước toàn bộ** (offline), có cách ngắn và nhanh hơn do dùng mảng liên tục (thân thiện với cache): **Fenwick Tree (BIT)** sau khi *nén tọa độ*, hoặc **sắp xếp trộn (merge sort)** đếm nghịch thế. Ô dưới cài Fenwick Tree: nén tọa độ để mỗi giá trị thành một chỉ số 1..m, `update(r)` cộng 1 tại vị trí `r`, `query(r)` cho tổng tần suất các vị trí 1..r. Số khóa nhỏ hơn `nums[i]` là `query(r - 1)`.

````python
def countSmaller_bit(nums):
    vals = sorted(set(nums))
    rank_of = {v: i + 1 for i, v in enumerate(vals)}   # nén tọa độ: giá trị -> 1..m
    m = len(vals)
    bit = [0] * (m + 1)

    def update(i):                                      # tần suất tại vị trí i += 1
        while i <= m:
            bit[i] += 1
            i += i & -i

    def query(i):                                       # tổng tần suất các vị trí 1..i
        s = 0
        while i > 0:
            s += bit[i]
            i -= i & -i
        return s

    res = [0] * len(nums)
    for i in range(len(nums) - 1, -1, -1):
        r = rank_of[nums[i]]
        res[i] = query(r - 1)                           # số phần tử đã thấy nhỏ hơn nums[i]
        update(r)
    return res


print(countSmaller_bit([5, 2, 6, 1]))

random.seed(3)
for _ in range(500):
    a = [random.randint(-20, 20) for _ in range(random.randint(1, 40))]
    assert countSmaller_bit(a) == countSmaller(a) == brute(a)
print("Fenwick, LLRB và vét cạn cho cùng kết quả trên 500 mảng ngẫu nhiên.")

# So sánh thời gian tương đối (phụ thuộc máy, chỉ mang tính tham khảo)
import time
big = [random.randint(-10**4, 10**4) for _ in range(20000)]
for name, f in [("LLRB", countSmaller), ("Fenwick", countSmaller_bit)]:
    t0 = time.perf_counter()
    f(big)
    print(f"{name:8s}: {time.perf_counter() - t0:.3f} giây (n = 20000)")
````

**Phân tích và khi nào cây cân bằng đáng dùng**

* Cả ba cách cho cùng kết quả, nên Fenwick là một phương án thay thế **đúng**.
* Thời gian ở ô trên đo trên máy chạy notebook này; **máy của bạn sẽ cho số khác**. Trong lần chạy tạo notebook này Fenwick nhanh hơn LLRB rõ rệt. Điều đáng so sánh là tỉ lệ giữa hai dòng, và nên chạy lại để lấy số liệu của chính máy mình khi thuyết trình. Fenwick dùng mảng liên tục và vòng lặp ngắn, còn LLRB phải cấp phát nút và đi theo con trỏ.
* **Fenwick có điều kiện:** cần biết trước toàn bộ giá trị để nén tọa độ. Khi dữ liệu đến liên tục (*online/streaming*), không biết trước miền giá trị, lại vừa chèn vừa **xóa** vừa tìm nhỏ nhất/lớn nhất, thì cây cân bằng (2-3, Red-Black, B-Tree) mới là lựa chọn tự nhiên, ví dụ bài **LeetCode 218 (The Skyline Problem)** dùng `TreeMap` làm multiset chiều cao với thêm, xóa, lấy khóa lớn nhất đều $\mathcal{O}(\log n)$, tổng $\mathcal{O}(n \log n)$ thay vì $\mathcal{O}(n^2)$ khi dùng danh sách.
* **Bài học:** chọn cấu trúc dữ liệu theo *tính chất dữ liệu và thao tác cần có*, chứ không phải cấu trúc "xịn" nhất.
